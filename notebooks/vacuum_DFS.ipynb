```python
class Node:
    def __init__(self, state, parent=None, action=None):
        self.state = state
        self.parent = parent
        self.action = action
```

```python
def is_goal(problem, state):
    return state == problem.goal_condition
```

```python
def Stack():
    return []
```

```python
def push(stack, node):
    stack.append(node)
```

```python
def is_empty(stack):
    return len(stack) == 0
```

```python
def pop(stack):
    if is_empty(stack):
        return None
    node = stack[-1]
    del stack[-1]
    return node
```

```python
def child_node(problem, parent, action):
    new_state = problem.result(parent.state, action)
    return Node(state=new_state, parent=parent, action=action)
```

```python
def solution(node):
    path = []
    while node is not None:
        path.append(node.state)
        node = node.parent
    path.reverse()
    return path
```

```python
class Problem:
    def __init__(self, initial_state, goal_condition, actions, result):
        self.initial_state = initial_state
        self.goal_condition = goal_condition
        self.actions = actions
        self.result = result
```

```python

def chi_so_phong(vi_tri):
    if vi_tri == "A":
        return 1
    else:
        return 2


def actions_vacuum(state):
    return ["Suck", "Left", "Right"]


def result_vacuum(state, action):
    new_state = state.copy()
    if action == "Suck":
        k = chi_so_phong(state[0])
        new_state[k] = "C"
    elif action == "Left":
        new_state[0] = "A"
    elif action == "Right":
        new_state[0] = "B"
    return new_state


def DFS(problem):
    node = Node(problem.initial_state)
    if is_goal(problem, node.state):
        return solution(node)
    frontier = Stack()
    push(frontier, node)
    explored = []
    while not is_empty(frontier):
        node = pop(frontier)
        explored.append(node.state)
        for action in reversed(problem.actions(node.state)):
            child = child_node(problem, node, action)

            in_frontier = False
            for n in frontier:
                if n.state == child.state:
                    in_frontier = True

            if child.state not in explored and not in_frontier:
                if is_goal(problem, child.state):
                    return solution(child)
                push(frontier, child)
    return None


def print_state(state):
    print("[" + ", ".join(state) + "]")
    print()


initial = ["A", "D", "D"]
goal = ["B", "C", "C"]
problem = Problem(initial, goal, actions_vacuum, result_vacuum)

path = DFS(problem)
if path is None:
    print("Không tìm thấy lời giải")
else:
    for i in range(len(path)):
        print("Bước", i)
        print_state(path[i])
    print("Số bước:", len(path) - 1)
```
